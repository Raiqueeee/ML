# Decision Tree Classifier

## Overview
This notebook trains a Decision Tree Classifier to predict the species of an iris flower from its measurements. It covers loading data, exploring it, splitting it, training the tree, and evaluating it with classification metrics and feature importances.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

## Dataset
The Iris dataset is built into scikit-learn (`sklearn.datasets.load_iris`). It has 150 samples, 4 features (sepal length, sepal width, petal length, petal width), and 3 classes: setosa, versicolor, virginica (50 samples each).

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Feature names:", list(X.columns))
print("Target names:", list(iris.target_names))
df = iris.frame
print(df.head())
```

```python
print(df.describe())
print("Class distribution:")
print(pd.Series(y).value_counts())
print("Missing values:", int(df.isnull().sum().sum()))
```

## Preprocessing and Train/Test Split
Decision trees split on feature thresholds, so no feature scaling is needed. The data is split 80/20 with stratification.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print("Training set:", X_train.shape, "Test set:", X_test.shape)
```

## Model Training
Train a Decision Tree Classifier. The depth is limited to 3 so the tree stays readable.

```python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
print("Tree depth:", model.get_depth(), "Number of leaves:", model.get_n_leaves())
```

## Evaluation
Predict on the test set and report accuracy, the classification report, and the confusion matrix.

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Visualization
Confusion matrix heatmap, feature importances, and the decision tree itself.

```python
plt.figure(figsize=(7, 5))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Greens',
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Decision Tree: Confusion Matrix')
plt.show()

importances = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(8, 5))
importances.plot(kind='bar')
plt.ylabel('Importance')
plt.title('Feature Importances (Decision Tree)')
plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 8))
plot_tree(model, feature_names=list(X.columns), class_names=list(iris.target_names), filled=True)
plt.title('Decision Tree')
plt.show()
```

## Conclusion
A shallow decision tree is easy to read and already separates the iris species well. Limiting the depth keeps the model simple and reduces overfitting.
