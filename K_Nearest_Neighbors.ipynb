# K-Nearest Neighbors (KNN)

## Overview
This notebook trains a K-Nearest Neighbors classifier to predict the species of an iris flower from its measurements. It covers loading data, exploring it, scaling the features, splitting the data, training the model, and evaluating it.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

## Dataset
The Iris dataset is built into scikit-learn (`sklearn.datasets.load_iris`). It has 150 samples, 4 features (sepal length, sepal width, petal length, petal width), and 3 classes: setosa, versicolor, virginica (50 samples each).

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
print("Features shape:", X.shape)
print("Feature names:", list(X.columns))
print("Target names:", list(iris.target_names))
df = iris.frame
print(df.head())
```

```python
print(df.describe())
print("Class distribution:")
print(pd.Series(y).value_counts())
print("Missing values:", int(df.isnull().sum().sum()))
```

## Preprocessing and Train/Test Split
KNN measures distances between points, so features must be on a similar scale. The features are standardised with StandardScaler (fitted on the training data only). The data is split 80/20 with stratification.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("Training set:", X_train.shape, "Test set:", X_test.shape)
```

## Model Training
Train a K-Nearest Neighbors classifier with k=5 neighbours.

```python
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train_scaled, y_train)
print("Model trained with k =", model.n_neighbors)
```

## Evaluation
Predict on the test set and report accuracy, the classification report, and the confusion matrix.

```python
y_pred = model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Visualization
Confusion matrix heatmap, and accuracy for different values of k to show how the choice of k affects the result.

```python
plt.figure(figsize=(7, 5))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Purples',
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('KNN (k=5): Confusion Matrix')
plt.show()

k_values = range(1, 16)
scores = []
for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    scores.append(accuracy_score(y_test, knn.predict(X_test_scaled)))
plt.figure(figsize=(8, 5))
plt.plot(list(k_values), scores, marker='o')
plt.xlabel('k (number of neighbours)')
plt.ylabel('Test accuracy')
plt.title('KNN Accuracy vs k')
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

## Conclusion
KNN is simple and effective on this small dataset once the features are scaled. The accuracy-vs-k plot shows that the result is stable across a range of k values.
