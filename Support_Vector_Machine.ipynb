# Support Vector Machine (SVM)

## Overview
This notebook trains a Support Vector Machine classifier to predict whether a breast tumour is malignant or benign. It covers loading data, exploring it, scaling the features, splitting the data, training the model, and evaluating it.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
```

## Dataset
The Breast Cancer Wisconsin dataset is built into scikit-learn (`sklearn.datasets.load_breast_cancer`). It has 569 samples, 30 features computed from cell nucleus measurements, and a binary target: malignant (0) or benign (1).

```python
cancer = load_breast_cancer(as_frame=True)
X = cancer.data
y = cancer.target
print("Features shape:", X.shape)
print("Target names:", list(cancer.target_names))
df = cancer.frame
print(df.head())
```

```python
print("Class distribution:")
print(pd.Series(y).value_counts())
print("Missing values:", int(df.isnull().sum().sum()))
print(df.describe().T[['mean', 'std', 'min', 'max']].head(5))
```

## Preprocessing and Train/Test Split
SVMs measure distances, so the features are standardised with StandardScaler (fitted on the training data only). The data is split 80/20 with stratification.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("Training set:", X_train.shape, "Test set:", X_test.shape)
```

## Model Training
Train a Support Vector Machine with the RBF kernel, the usual default for this kind of problem.

```python
model = SVC(kernel='rbf', random_state=42)
model.fit(X_train_scaled, y_train)
print("Number of support vectors per class:", model.n_support_)
```

## Evaluation
Predict on the test set and report accuracy, the classification report, and the confusion matrix.

```python
y_pred = model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
print(classification_report(y_test, y_pred, target_names=cancer.target_names))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Visualization
Confusion matrix heatmap.

```python
plt.figure(figsize=(7, 5))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Oranges',
            xticklabels=cancer.target_names, yticklabels=cancer.target_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('SVM (RBF kernel): Confusion Matrix')
plt.show()
```

## Conclusion
An SVM with the RBF kernel separates malignant from benign cases very well on this dataset once the features are scaled. Scaling is essential here because the kernel uses distances.
